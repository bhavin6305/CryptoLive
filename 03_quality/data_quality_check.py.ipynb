```python
# Config from ../04_utils/shared_config (inlined because %run path resolution fails on serverless)
CATALOG = "workspace"
SCHEMA = "crypto_live"

fact_count = spark.table(f"{CATALOG}.{SCHEMA}.fact_crypto_prices").count()
assert fact_count > 0, "fact_crypto_prices is empty!"

null_prices = spark.table(f"{CATALOG}.{SCHEMA}.fact_crypto_prices").filter("price_usd IS NULL").count()
print(f"Rows with missing price_usd: {null_prices}")

# Sanity check: no coin should have a negative price, ever
negative_prices = spark.table(f"{CATALOG}.{SCHEMA}.fact_crypto_prices").filter("price_usd < 0").count()
assert negative_prices == 0, f"Found {negative_prices} rows with negative prices — data corruption!"

print(f"All checks passed. Total fact table rows: {fact_count}")
```
